```python
import requests
```

```python
basic_url = 'https://gorest.co.in/'
end_point = 'public/v2/users'
# parameters = '?page=1&per_page=100'
full_url = basic_url + end_point

r = requests.get(full_url, params= {'page' : 1, 'per_page' : 100})
r
```

```python
full_url
```

```python
r.status_code
```

```python
data = r.json()
```

```python
import csv
```

```python
for d in data:
    print(d['name'])
```

```python
with open('users.csv', 'w', newline= '') as file:

    writer = csv.DictWriter(file, ['id', 'name', 'email', 'gender', 'status'])
    writer.writeheader()

    for d in data:
        id = d['id']
        name = d['name']
        email = d['email']
        gender = d['gender']
        status = d['status']

        writer.writerow({'id' : id, 'name' : name, 'email' : email, 'gender' : gender, 'status' : status})
```
