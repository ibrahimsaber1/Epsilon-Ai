```python
from bs4 import BeautifulSoup
import requests
import csv
```

```python
page_number = 0

with open('Wuzzuf.csv', 'w', newline= '') as file:

    writer = csv.DictWriter(file, fieldnames= ['Job title', 'Company', 'Job type', 'Skills'])
    writer.writeheader()

    while True:
        response = requests.get(f'https://wuzzuf.net/search/jobs/?q=data%20science&a=hpb&start={page_number}')
        data = BeautifulSoup(response.text, 'html.parser')
        
        if data.find_all('div', 'css-pkv5jc') == []:
            print('Done Scraping All Pages')
            break

        for job in data.find_all('div', 'css-pkv5jc'):
            try:
                job_title = job.div.h2.a.text
                company = job.div.div.a.text
                job_type = job.find('div', 'css-y4udm8').div.a.span.text
                skills_list = job.find('div', 'css-y4udm8').find_all('a', 'css-5x9pm1')
                skills = [skill.text.split('·')[1].strip() for skill in skills_list]
                writer.writerow({'Job title' : job_title, 'Company' : company, 'Job type' : job_type, 'Skills' : skills})
            except :
                pass
        page_number += 1
        
        if page_number == 50:
            print('Done Scraping All Pages')
            break
```
