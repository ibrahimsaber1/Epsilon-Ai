```python
page_number = 1
with open('Diwan.csv', 'w', newline= '') as file:

    writer = csv.DictWriter(file, fieldnames= ['Title', 'Price', 'Image'])
    writer.writeheader()

    while True:
        response = requests.get(f'https://diwanegypt.com/product-category/books/english-adults/page/{page_number}')
        data = BeautifulSoup(response.text, 'html.parser')

        if data.find_all('a', 'woocommerce-LoopProduct-link woocommerce-loop-product__link') == []:
            print('Done Scraping All Pages')
            break

        for book in data.find_all('a', 'woocommerce-LoopProduct-link woocommerce-loop-product__link'):

            Title = book.h2.text
            Price = book.bdi.text
            Image = book.img['src']

            writer.writerow({'Title' : Title, 'Price' : Price, 'Image' : Image})

        page_number += 1
        print(f'Page {page_number} Scraped')
        if page_number == 50:
            print('Done Scraping All Pages')
            break
```
