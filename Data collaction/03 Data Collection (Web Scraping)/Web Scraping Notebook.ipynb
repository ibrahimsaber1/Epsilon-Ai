```python
from bs4 import BeautifulSoup
import requests
import csv
```

```python
response = requests.get('https://books.toscrape.com/')
response
```

```python
data = BeautifulSoup(response.text, 'html.parser')
data
```

```python
with open('books.csv', 'w', newline= '') as file:

    writer = csv.DictWriter(file, fieldnames= ['Name', 'Price', 'Stock Availability'])
    writer.writeheader()

    for book in data.find_all('article', 'product_pod'):
        book_name = book.h3.a['title']
        book_price = book.find_all('div')[1].p.text[2:]
        book_availability = book.find_all('div')[1].find_all('p')[1].text.strip()

        writer.writerow({'Name' : book_name, 'Price' : book_price, 'Stock Availability' :book_availability})
```

```python
response = requests.get('https://wuzzuf.net/search/jobs/?q=data%20science&a=hpb')
response
```

```python
data = BeautifulSoup(response.text, 'html.parser')
data
```

```python
with open('Wuzzuf.csv', 'w', newline= '') as file:

    writer = csv.DictWriter(file, fieldnames= ['Job title', 'Company', 'Job type', 'Skills'])
    writer.writeheader()

    for job in data.find_all('div', 'css-pkv5jc'):
        try:
            job_title = job.div.h2.a.text
            company = job.div.div.a.text
            job_type = job.find('div', 'css-y4udm8').div.a.span.text
            skills_list = job.find('div', 'css-y4udm8').find_all('a', 'css-5x9pm1')
            skills = [skill.text for skill in skills_list]
            writer.writerow({'Job title' : job_title, 'Company' : company, 'Job type' : job_type, 'Skills' : skills})
        except:
            pass
```
