```python
from sklearn.impute import KNNImputer, SimpleImputer
# the logic of the KNNImputer is to fill the missing values with the mean of the k nearest neighbors
# n_neighbors is the number of neighbors to consider for filling the missing values
# n_neighbors=5 means that the mean of the 5 nearest neighbors will be used to fill the missing values 

knn = KNNImputer(n_neighbors= 5)

df['BuildingArea'] = knn.fit_transform(df[['BuildingArea']])
df['BuildingArea']
```

```python
# The SimpleImputer is used to fill the missing values with a constant value or the mean/median of the column
# as u see i first creat an instance of the SimpleImputer class
sim_impu = SimpleImputer(strategy= 'median')

df[['Car', 'YearBuilt']] = sim_impu.fit_transform(df[['Car', 'YearBuilt']])
```

```python
from sklearn.preprocessing import OneHotEncoder

ohe = OneHotEncoder(drop= 'first', sparse_output= False)

ohe_arr = ohe.fit_transform(df_cat[['Method']])
ohe_arr
```
