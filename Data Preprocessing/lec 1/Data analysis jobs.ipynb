# Data Cleaning

```python
df.dropna(inplace= True, ignore_index= True)
```

```python
df
```

```python
# THIS will replace -1 and '-1' with NaN
df = df.replace('-1', np.nan).replace(-1, np.nan)
df
```

```python
df.isna().sum()
```

```python
df.isna().mean().round(4) * 100
```

```python
df.dropna(subset= ['Salary Estimate'], inplace= True, ignore_index= True)
```

```python
# drop the 'Competitors' and 'Easy Apply' columns becouse the null values in these columns above 40%
df.drop(['Competitors', 'Easy Apply'], axis= 1, inplace= True)
```

```python
df.duplicated().sum()
```

```python
cat_cols = df.select_dtypes(include= 'object').columns
cat_cols
```

```python
for col in cat_cols:
    print(col)
    print(df[col].nunique())
    print(df[col].unique())
    print('\n', '-' * 100, '\n')
```

```python
with open('cat_cols.txt','w') as f:
    for col in cat_cols:
        f.write(f"{col}\n")
        f.write(f"Unique values: {df[col].nunique()}\n")
        f.write(f"Values:\n {df[col].unique()}\n")
        f.write('\n' + '-*' * 100 + '\n')
```

```python
df['Company Name']= df['Company Name'].str.split("\n").str[0]
df['Company Name']
```

```python
df.head()
```

```python
df['min_salary'] = (df['Salary Estimate'].str.split('-').str[0].str[1:3] + '000').astype(float)
df['min_salary']
```

```python
df['max_salary'] = (df['Salary Estimate'].str.split('-').str[1].str.split().str[0].str[1:3] + '000').astype(float)
df['max_salary']
```

```python
df['avg_salary'] = (df.min_salary + df.max_salary) / 2
df['avg_salary']
```
