# Handling Outliers

## 1) Using datasist library.

```python
import seaborn as sns
df = sns.load_dataset('tips')
df.head()
```

```python
px.box(df, y= 'total_bill')
```

### Detect Outlier Indices

```python
from datasist.structdata import detect_outliers

outlier_indices = detect_outliers(df, 0, ['total_bill'])
outlier_indices
```

### Replace Outlier Indices with the Median

```python
from datasist.structdata import detect_outliers

outlier_indices = detect_outliers(df, 0, ['total_bill'])

Median = df.total_bill.median()

df.total_bill.loc[outlier_indices] = Median
```

## 2) Using Custom Function.

```python
def custom_outliers(dataframe, column, lower_bound = 0.1, upper_bound = 0.9):

    # Compute percentile boundaries
    lower_threshold = dataframe[column].quantile(lower_bound)
    upper_threshold = dataframe[column].quantile(upper_bound)

    # Find outlier indices
    outlier_indices = df[(dataframe[column] < lower_threshold) | (dataframe[column] > upper_threshold)].index
    return outlier_indices
```

```python
# Detect outliers lower than 20% value or higher than 80% value
custom_outliers(dataframe= df, column= 'total_bill', lower_bound= 0.2, upper_bound= 0.8)
```

```python
# Detect outliers lower than 10% value or higher than 90% value
custom_outliers(dataframe= df, column= 'total_bill', lower_bound= 0.1, upper_bound= 0.9)
```
