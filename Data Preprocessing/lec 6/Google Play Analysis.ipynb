```python
def clean_size(x):

    if type(x) == float:
        return x
    
    elif x == 'Varies with device':
        return x
    
    elif 'M' in x:
        return x
    
    elif 'k' in x:
        return str(round(float(x[:-1]) / 1024, 1)) + 'M'
    
    else:
        return x
    
df['Size'] = df['Size'].apply(clean_size)
df['Size'] 
```

```python
str(round(float('498k'[:-1]) / 1024, 1)) + 'M'
```

```python
df.Installs.unique()
```

```python
def clean_installs(x):

    if x in ['10+', '1+', '5+', '0+', '0', '50+', '100+', '500+']:
        return '< 1000'
    
    elif x in ['1,000+', '5,000+', '50,000+', '10,000+']:
        return '< 100000'
    
    elif x in ['100,000+', '500,000+']:
        return '< 1000000'
    
    else:
        return '>= 1000000'
    
df.Installs = df.Installs.apply(clean_installs)
df.Installs
```

```python
df.Price = df.Price.str.replace('$', '').astype(float)
df.Price
```

```python
%%writefile app.py

import pandas as pd
import streamlit as st
import plotly.express as px

st.set_page_config(layout= 'wide', page_title= 'Goodle Play EDA')

# st.title('Google Play Exploratory Data Analysis')

st.markdown("<h1 style='text-align: center; color: white;'>Google Play Exploratory Data Analysis</h1>", unsafe_allow_html=True)

st.image('https://cdn.freebiesupply.com/logos/large/2x/google-play-badge-1-logo-png-transparent.png')

df = pd.read_csv('cleaned_df.csv', index_col= 0)

page = st.sidebar.radio('Pages', ['Introduction', 'Analysis Questions', 'Reporting'])

if page == 'Introduction':

    st.dataframe(df.head())

    st.header('Data Description')

    st.write('''Category : Category the app belongs to

Rating : Overall user rating of the app (as when scraped)


Reviews : Number of user reviews for the app (as when scraped)


Size : Size of the app (as when scraped)


Installs : Number of user downloads/installs for the app (as when scraped)


Type : Paid or Free


Price : Price of the app (as when scraped)


Content Rating : Age group the app is targeted at - Children / Mature 21+ / Adult


Genres : An app can belong to multiple genres (apart from its main category). For eg, a musical family game will belong to Music, Game, Family genres.''')

elif page == 'Analysis Questions':

    st.header('What is the most expensive App on the store ?')
    st.write(df.sort_values(by= 'Price', ascending= False)[['App', 'Price']].head(1))

    st.header('How many applications that has more than 50k reviews')

    if st.button('Check answer'):
        st.write(df[df.Reviews > 50000]['App'].nunique())

    st.header('What is the Average Reviews per Content Rating per Type ?')
    content_type_review = df.groupby(['Content Rating', 'Type'])['Reviews'].mean().reset_index()
    st.plotly_chart(px.bar(data_frame= content_type_review, x= 'Content Rating', y= 'Reviews', 
       labels= {'Reviews' : 'Average No of Reviews'}, title= 'Average Reviews per Content Rating per Type',
       text_auto= True, color= 'Type', barmode= 'group'))

elif page == 'Reporting':

    year = st.sidebar.selectbox('Year', df['last_updated_year'].unique())

    category = st.sidebar.selectbox('Category', df['Category'].unique())

    df_2 = df[(df['last_updated_year'] == year) & (df['Category'] == category)]

    st.dataframe(df_2)
```
