```python
def clean_invoicedate(x):

    x = x.split('/')
    x[-1] = '20' + x[-1]
    x = '/'.join(x)
    return x

df.invoicedate = df.invoicedate.apply(clean_invoicedate)
df.invoicedate
```

```python
cat_count = df.groupby('major_category')['invoiceno'].count().sort_values(ascending= False).reset_index()
cat_count
```

```python
px.pie(cat_count, names= 'major_category', values= 'invoiceno', title= 'Total Orders Per Category')
```

```python
# get the total order value per month

cat_month = df.groupby('month')['ordervalue'].sum().sort_values(ascending= False).reset_index()
cat_month
```

```python
px.bar(cat_month, x = 'month', y= 'ordervalue')
```
