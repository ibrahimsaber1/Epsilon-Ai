## Import libraries

```python
import pandas as pd
import numpy as np
```

## Load data

```python
customers = pd.read_csv('./Shopping Cart Dataset/customers.csv')
customers
```

```python
orders = pd.read_csv('./Shopping Cart Dataset/orders.csv')
orders
```

```python
products = pd.read_csv('./Shopping Cart Dataset/products.csv')
products
```

```python
sales = pd.read_csv('./Shopping Cart Dataset/sales.csv')
sales
```

## Merge the data

```python
df = pd.merge(customers, orders, how= 'inner', on= 'customer_id')
df
```

```python
df = pd.merge(df, sales, how= 'inner', on= 'order_id')
df
```

```python
df = pd.merge(df, products, how= 'inner', left_on= 'product_id', right_on= 'product_ID')
df
```

### Data Cleaning

```python
# convert to datetime format
df['order_date'] = pd.to_datetime(df['order_date'])

df['delivery_date'] = pd.to_datetime(df['delivery_date'])
```

```python
df.info()
```

```python
df['order_quantity'] = df.quantity_x

df['stock_quantity'] = df.quantity_y

df = df.drop(['quantity_x', 'quantity_y'], axis= 1)
df.head()
```

```python
# extract year , month and day name for order data
df['order_year'] = df.order_date.dt.year

df['order_month'] = df.order_date.dt.month

df['order_day'] = df.order_date.dt.day_name()

df.head(1)
```

```python
df.order_date.dt.day_name()
```

```python
# extract year , month and day name for delivery data
df['delivery_year'] = df.delivery_date.dt.year

df['delivery_month'] = df.delivery_date.dt.month

df['delivery_day'] = df.delivery_date.dt.day_name()

df.head(1)
```

```python
# what is the total money that each gendre spent
df.groupby('gender')['payment'].sum().sort_values(ascending= False)
```

```python
# what is the most day that get highest revenue
df.groupby('order_day')['payment'].sum().sort_values(ascending= False)
```

```python
df.head(1)
```

```python
# what is the highest month that get the highest sales?
df.groupby('order_month')['order_id'].count().sort_values(ascending= False)
```

```python
# what is the highest state achieve the highest salees
df.groupby('state')['order_id'].count().sort_values(ascending= False)
```

```python
# what is the highest state per gender achieve the highest sales
df.groupby(['state', 'gender'])['order_id'].count().head(20)
```

```python
df.pivot_table('order_id', index= 'state', columns= 'gender', aggfunc= 'count')
```

```python
# what is the average age per state
df.groupby('state')['age'].mean().sort_values(ascending= False)
```

```python
# what is the average age per state per each gender
df.groupby(['state', 'gender'])['age'].mean().head(20)
```

```python
df.pivot_table(values= 'age', index= 'state', columns= 'gender')
```

```python
# what is the product type that have the highest revenue
df.groupby('product_type')['payment'].max().sort_values(ascending= False)
```
