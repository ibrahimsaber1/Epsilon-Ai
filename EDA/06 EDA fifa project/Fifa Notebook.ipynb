# Data Cleaning

```python
# Make columns lowercase and replace space with underscore
df.columns = df.columns.str.lower().str.replace(' ', '_')
df.columns
```

```python
# Update datatype of contract column
df.contract_valid_until = pd.to_datetime(df.contract_valid_until)
df.contract_valid_until
```

- Check inconsistent minimum value in wage column which is equal to zero

```python
df[df.wage < 1]
```

```python
# Detect all rows that need to be dropped
drop_indices = df[df.wage < 1].index
drop_indices
```

```python
len(drop_indices)
```

```python
# Drop all rows where wage equal 0
df.drop(drop_indices, inplace= True)
```

```python
df
```

```python
# Reset the index always after removing any rows from the data
df.reset_index(drop= True, inplace= True)
```

```python
df
```

### Handling Missing Values

```python
round((df.isna().sum() / df.shape[0]) * 100, 3)
```

```python
df.value
```

```python
# replace all missing values in value column with the median
value_median = df.value.median()
value_median
```

```python
df.value = df.value.fillna(value_median)
df.value
```

```python
round((df.isna().sum() / df.shape[0]) * 100, 3)
```

```python
df[ (df.international_reputation.isna()) & (df.skill_moves.isna()) & (df.contract_valid_until.isna()) ].head()
```

```python
# Drop all Missing Values in the dataframe
df.dropna(inplace= True, ignore_index= True)
```

### In depth check for categorical columns

```python
cat_cols = df.select_dtypes(include= 'object').columns
cat_cols
```

```python
with open('categorical_columns.txt', 'w') as f:
    for col in cat_cols:
        f.write(f'{col} :\n{df[col].nunique()}\n{df[col].unique()}\n\n')
        f.write('--------------------------------\n')
for col in cat_cols:
    print(col)
    print(df[col].nunique())
    print(df[col].unique(), '\n')
```

### In depth check for numerical columns

```python
num_cols = df.select_dtypes(include= ['int', 'float']).columns
num_cols
```

```python
for col in num_cols:
    px.histogram(df[col]).show()
```

```python
# Convert Height unit from feet to cm
df.height = df.height.apply(lambda x : x * 30.48)
df.height
```

```python
# Convert Weight unit from pound to kilogram
df.weight = df.weight.apply(lambda x : x * 0.453592)
df.weight
```

### **What is the average overall rating for each country ?**

```python
top_10_overall = df.groupby('nationality')['overall'].mean().sort_values(ascending= False).head(10)
top_10_overall
```

```python
type(top_10_overall)
```

```python
top_10_overall.index
```

```python
top_10_overall.values
```

```python
px.bar(top_10_overall, x = top_10_overall.index, y= top_10_overall.values, labels= {'nationality' : 'Nationality', 'y' : 'Average Overall'})
```

```python
df.columns
```

### Another way to plot the same Chart

```python
top_10_overall = df.groupby('nationality')['overall'].mean().sort_values(ascending= False).head(10).reset_index()
top_10_overall
```

```python
px.bar(top_10_overall, x = 'nationality', y= 'overall', title= 'Top 10 Countries per Overall',
                                                        labels= {'nationality' : 'Nationality', 'overall' : 'Average Overall'})
```

### **What are the top 10 clubs with highest number of playes ?**

```python
top_10_clubs = df.groupby('club')['id'].count().sort_values(ascending= False).head(10).reset_index()
top_10_clubs
```

```python
px.bar(top_10_clubs, x = 'club', y= 'id', title= 'Top 10 Countries per number of players', labels= {'club' : 'Club', 'id' : 'Count'})
```

### **what are the Top 20 clubs with highest player Value ?**

```python
top_club_value = df.groupby('club')['value'].sum().sort_values(ascending= False).reset_index().head(20)
top_club_value
```

```python
px.bar(top_club_value, x= 'club', y= 'value', title= 'Top 10 Clubs with Highest Player Values',
                                                        labels= {'club' : 'Club', 'value' : 'Total Player Value'})
```

### **What is the average height per position ?**

```python
avg_position_height = df.groupby('position')['height'].mean().sort_values(ascending= False).reset_index()
avg_position_height
```

```python
px.bar(avg_position_height, x= 'position', y= 'height', title= 'Average height per position',
                                                        labels= {'position' : 'Position', 'height' : 'Average Height'})
```

### **What is the average weight per position ?**

```python
avg_position_weight = df.groupby('position')['weight'].mean().sort_values(ascending= False).reset_index()
avg_position_weight
```

```python
px.bar(avg_position_weight, x= 'position', y= 'weight', title= 'Average Weight per position',
                                                        labels= {'position' : 'Position', 'weight' : 'Average Weight'})
```
