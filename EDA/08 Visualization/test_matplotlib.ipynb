```python
import matplotlib.pyplot as plt
import plotly.express as px
```

```python
nvals = []
linear = []
quadratic = []
cubic = []
exponential = []

for n in range(1, 30):
    nvals.append(n)
    linear.append(n)
    quadratic.append(n**2)
    cubic.append(n**3)
    exponential.append(2**n)
    
print("nvals", nvals)
print("linear", linear)
print("quadratic", quadratic)
print("cubic", cubic)
print("exponential", exponential)
```

```python
plt.plot(nvals, linear, label='test lable')
```

```python
plt.bar(nvals, linear, label='linear', alpha=0.5)
```
