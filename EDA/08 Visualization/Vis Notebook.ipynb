```python
import plotly.express as px
```

```python
px.data.__all__
```

```python
df = px.data.tips()
df
```

```python
px.histogram(data_frame= df, x = 'total_bill')
```

```python
px.histogram(df.smoker)
```

```python
df.day.value_counts()
```

```python
day_values = df.day.value_counts().values
day_values
```

```python
day_categories = df.day.unique()
day_categories
```

```python
df.day.value_counts()
```

```python
px.pie(data_frame= df, names= day_categories, values= day_values)
```

```python
# Total Bill per day

bill_per_day = df.groupby('day')['total_bill'].sum().sort_values(ascending= False).reset_index()
bill_per_day
```

```python
px.bar(bill_per_day, x= 'day', y= 'total_bill', title= 'Total Bill per Day',
              labels= {'total_bill' : 'Total Bill Amount'})
```

### Numerical vs Numerical
