## 1) Arithmetic and Logic

You can easily perform array with array arithmetic, or scalar with array arithmetic. <br>
You can also perform logic operation on array. <br>

**Broadcasting**: a technique in numpy used to make operations on different array sizes like multiply or sum a scaler to an array.

Let's see some examples:

```python
import numpy as np
```

### 1.1) Arithmetic

```python
# concate two lists will reture a list have all the elements of both lists
l1 = [1, 2, 3, 4, 5]
l2 = [7, 8, 9, 10, 11]
l1 + l2 # [1, 2, 3, 4, 5, 7, 8, 9, 10, 11]
# zip two lists will return a list of tuples, each tuple contains one element from each list
l3 = []
for n1, n2 in zip(l1, l2):
    l3.append(n1 + n2)

l3
```

```python
# concat two numpy arrays will return a new array with summation of each element of both arrays
l1_arr = np.array([1, 2, 3, 4, 5])
l2_arr = np.array([7, 8, 9, 10, 11])
l1_arr + l2_arr
```

```python
# divide any number by 0 will return ZeroDivisionError
try :
    4 / 0
except ZeroDivisionError as e:
    print(" this is a ZeroDivisionError")
```

```python
0 / 0
```

```python
# Warning on division by zero, but not an error!
# Just replaced with nan

arr = np.array([0, 2, 3, 4, 5])
print(l2_arr, "/", arr)
l2_arr / arr
```

```python
arr = np.array([0, 2, 3, 4, 5])
arr_0 = np.array([0, 5, 7, 8, 9])
arr_0 / arr
```

### 2.4) dot product

```python
matrix * matrix
```

```python
arr_1 = np.array([ [1, 3, 5] ,[2, 4, 6] ])
arr_2 = np.array([ [2, 4] , [5, 6], [3, 1] ])

np.matmul(arr_1, arr_2)
```

```python
arr_1 = np.array([ [1, 3, 5, 3] ,[2, 4, 6, 4] ])
arr_2 = np.array([ [2, 4] , [5, 6], [3, 1] ])

np.matmul(arr_1, arr_2)
```
