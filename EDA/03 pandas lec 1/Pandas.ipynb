# **Series**
####  A Series is very similar to a NumPy array (in fact it is built on top of the NumPy array object). What differentiates the NumPy array from a Series, is that a Series can have axis labels, meaning it can be indexed by a label, instead of just a number location.
* A Pandas Series is like a column in a table.
* It is a one-dimensional array holding data of any type.

### 1) Creating a Series

#### Using list

```python
my_list = [10,20,30, 40, 50]
ser1 = pd.Series(my_list)
ser1
```

```python
# Using a list with customized index (add parameter index)
ser2 = pd.Series(my_list, index=['a', 'b', 'c', 'd', 'e'])
ser2
```

#### Using Dictionary

```python
# Using a dictionary (keys will be transformed into Series index and values will be the column values)
my_dict = {'a': 10, 'b': 20, 'c': 30, 'd': 40, 'e': 50}
ser3 = pd.Series(my_dict)
ser3
```

#### Using Numpy Array

```python
# Using a Numpy Array
import numpy as np

arr = np.array([10, 20, 30, 40, 50])
ser4 = pd.Series(arr)
ser4
```

### 2) Access Series Values

#### By Index

```python
print(ser4[0])
print(ser4[1])
print(ser4[2])
print(ser4[3])
```

#### By Slicing

```python
# All values
print(ser4[:])
# Range of values
print(ser4[0:2])
# Values reversed
print(ser4[::-1])
```

### 3) Series Operations

```python
ser5 = pd.Series(np.arange(1, 11))
ser6 = pd.Series(np.arange(10, 20))
print(ser5)
print(ser6)
```

#### Addition

```python
ser5 + ser6
```

#### Subtraction

```python
ser6 - ser5
```

#### Multiplication

```python
ser5 * ser6
```

#### Division

```python
ser6 / ser5
```

### 4) Methods

#### Useful Statistical Methods

```python
ser7 = pd.Series([1, 1, 3, 7, 1, 4, 5, 3, 9, 3])
ser7
```

```python
print('the ser7 is ==>',ser7)
print('*--'*20)
print('the summation of ser7 is ==>',ser7.sum())
print('the mean of ser7 is ==>',ser7.mean())
print('the min of ser7 is ==>',ser7.min())
print('the max of ser7 is ==>',ser7.max())
print('the count of ser7 is ==>',ser7.count())
print('the std of ser7 is ==>',ser7.std())
print('the var of ser7 is ==>',ser7.var())
print('the median of ser7 is ==>',ser7.median())
print('the mode of ser7 is ==>',ser7.mode())
```

#### Mode

```python
ser8 = pd.Series(['Egypt', 'USA', 'UK', 'Egypt', 'Egypt', 'Germany', 'UK', 'France', 'UK'])
print(ser8)
```

```python
# The mode method returns a series of all modes in the data
print(ser8.mode())                                         # Here we have two modes Egypt and UK
```

```python
# To get the First mode
print(ser8.mode()[0])
```

```python
# To get the Second mode
print(ser8.mode()[1])
```

# **DataFrame**
* A Pandas DataFrame is a 2 dimensional data structure, like a 2 dimensional array, or a `table` with rows and columns.
* Index represents rows

### 1) Creating DataFrame



#### Using Numpy Array

```python
df1 = pd.DataFrame(np.random.randn(5,4))          # Create a 5 X 4 two dimensional array of random numbers
df1
```

```python
# Using array with customized index (add parameter index)
df2 = pd.DataFrame(np.random.randn(5,4), index=['a', 'b', 'c', 'd', 'e'])
df2
```

```python
# Using array with customized index (add parameter index) and customized column names (add columns parameter)
df3 = pd.DataFrame(np.random.randn(5,4), index=['a', 'b', 'c', 'd', 'e'], columns=['col1', 'col2', 'col3', 'col4'])
df3
```

#### Using Dictionary

```python
# create dataframe Using Dictionary
my_dict = {'col1': [1, 2, 3, 4, 5], 'col2': [6, 7, 8, 9, 10], 'col3': [11, 12, 13, 14, 15]}
df4 = pd.DataFrame(my_dict)                # Here we don't have to add columns parameter because it takes the column names from dict keys
df4
```

### 2) Select, Create and Drop Columns

#### Select column



```python
# Select column
print(df4['col3'])
```

```python
# Select a list of columns
print(df4[['col1', 'col2']])
```

```python
df4
```

#### Create column

```python
# Create column using list
df4['col4'] = [100, 200, 300, 400, 500]         # The same way as adding a new key-value to a dictionary
df4
```

```python
# Create column using Numpy Array
df4['col5'] = np.arange(1, 6)
df4
```

```python
# Create column using Series
df4['col6'] = pd.Series([10, 20, 30, 40, 50])
df4
```

#### Drop Column

```python
# Drop column from Dataframe (Without UPDATING the variable)
df4.drop('col6', axis = 1)
```

```python
# Drop column from Dataframe (With UPDATING the variable, add inplace parameter)
df4.drop('col6', axis = 1, inplace = True)
```

```python
# Or we can use traditional way to update the variable
df4 = df4.drop('col5', axis = 1)
df4
```

### 3) Select Create and Drop Rows

#### Select Row

```python
df5 = pd.DataFrame(np.random.randn(5,4), index=['a', 'b', 'c', 'd', 'e'], columns=['col1', 'col2', 'col3', 'col4'])
df5
```

```python
# Select row (using iloc method)
df5.iloc[0]                                    # This method needs the index of the row starting from 0
```

```python
# Select a list of rows (using iloc method)
df5.iloc[0:3]
```

```python
# Select specific rows (using iloc method)
df5.iloc[[0, 3]]
```

```python
# Select row from dataframe (using loc method)
df5.loc['a', 'col1']                                    # This method needs the name of the index
```

```python
# Select a list of rows (using loc method)
df5.loc['a':'c']
```

```python
# Select specific rows (using loc method)
df5.loc[['a', 'd']]
```

#### Create Row

```python
# Add a new row with index f
df5.loc['f'] = [1, 2, 3, 4]
df5
```

```python
# Update an existing row
df5.loc['f'] = [100, 200, 300, 400]
df5
```

#### Drop Row

```python
# Drop Row (we use the same method for dropping columns but make axis = 0)
df5.drop('f', axis = 0)
```

```python
# To drop the row and UPDATE the variable
df5.drop('f', axis = 0, inplace = True)
df5
```

### 4) DataFrame Mask

```python
df6 = pd.DataFrame({'name': ['ahmed', 'mohamed', 'ali', 'ziad', 'yasser'],
                    'salary': [1000, 2000, 3000, 4000, 5000],
                    'age': [20, 25, 30, 35, 40]})
df6
```

#### Create Mask

```python
# A mask is any filter applied to the dataframe
mask1 = df6['salary'] > 2000                   # Here we need to get only rows where salary is greater than 2000
mask1                                          # The output is True and False because we used the comparison operator `>`
```

```python
# To get the values corresponding to True
df6[mask1]
```

#### Mask Logical Operators

```python
# To use the logical operator `AND` in the mask we replace it with `&` ( Make sure to put each condition in parenthesis() )
mask2 = (df6['salary'] > 2000) & (df6['age'] > 30)
mask2
```

```python
df6[mask2]
```

```python
# To use the logical operator `OR` in the mask we replace it with `|` ( Make sure to put each condition in parenthesis() )
mask3 = (df6['salary'] > 2000) | (df6['age'] > 30)
mask3
```

```python
df6[mask3]
```

```python
# To use the logical operator `NOT` in the maske we replace it with `~` TELDA
mask4 = ~(df6['salary'] > 2000)
mask4
```

```python
df6[mask4]
```
