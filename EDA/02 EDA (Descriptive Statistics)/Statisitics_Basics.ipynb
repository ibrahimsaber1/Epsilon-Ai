# Measuers of Center (Central Tendency of Data)

## **1- Mean (Average)** :
#### Found by adding all numbers in the data set and then dividing by the number of values in the set.
# $\mu = \frac{\sum_{i=1}^n (x_i)}{n}$

```python
# Calculate mean with Traditional way

data = np.array([10, 12, 13, 0, 4, -5, 2])
data_mean = sum(data) / len(data)
print(data_mean)

# Calculate mean using numpy
print(np.mean(data))
```

```python
arr01 = np.array([15,80,9,6,14,2,3,4,5,6])
# Calculate mean of two arrays
arr_mean = sum(arr01) / len(arr01)
print(arr_mean)
# Calculate mean of two arrays using numpy
arr_mean_np = np.mean(arr01)
print(arr_mean_np)
```

## **2- Median** :
#### The median is the middle value in a `Sorted` data

image.png

```python
# Calculate Data Median for odd number of values
data = np.array([10, 12, 13, 0, 4, -5, 2])

data.sort()
print(f'Sorted data : {data}')
median_idx = len(data) // 2
median_odd = data[median_idx]
print(median_odd)

# Calculate Median using numpy in one step
print(np.median(data))
```

```python
# Calculate Data Median for even number of values
data =  np.array([10, 12, 13, 0, 4, -5, 2, 12])          # Here we added 12 at the end to make the number of elements even
data.sort()
print(f'Sorted data : {data}')
first_median_idx = len(data) // 2
second_median_idx = first_median_idx - 1
median_even = (data[first_median_idx] + data[second_median_idx]) / 2
print(median_even)

# Caclulate Median using numpy in one step
print(np.median(data))
```

# Measures of Spread
### Used to descripe the variation and spread in data

## What’s the difference between central tendency and dispersion (spread) ?
* While central tendency tells you where most of your data points lie, dispersion summarizes how far apart your points from each other.

- Data sets can have the same central tendency but different levels of dispersion or vice versa. Together they give you a complete picture of your data

## **1- Range**
#### The difference between the Maximum and Minimum value in data
# $Range = x_{max} - x_{min}$

```python
data =  np.array([10, 12, 13, 7, 4, 5, 2, 12])
range = max(data) - min(data)
print(f'The Maximum value in the data = {max(data)}')
print(f'The Minimum value in the data = {min(data)}')
print(f'The Range of data = {range}')
```

```python
data =  np.array([10, 12, 13, 7, 4, 5, 2, 12])
# calculate range using numpy
data_range = np.ptp(data)
print(f'The Range of data using numpy = {data_range}')
```

## **2- Inter Quartile Range (IQR)**
#### It is a better measure of spread than range because it leaves out the extreme values.
# $IQR = Q3 - Q1 $

image.png

### Advantage of IQR:
* The main advantage of the IQR is that it is not affected by outliers because it doesn’t take into account observations below Q1 or above Q3
* observations can be outliers when they lie more than 1.5 IQR below the first quartile or 1.5 IQR above the third quartile.
* Outliers = Q1 –1.5* IQR (or) Q3 + 1.5*IQR

## **3- Variance**
#### A measure of how far a group of data is spread out from their mean.
# $\sigma^2 = \frac{\sum_{i=1}^n (x_i - \mu)^2}{n}$
* $\mu$ : mean of data
- n : number of data points

image.png

### How to calculate variance step by step
#### 1- Calculate the mean : $\mu$
#### 2- Subtract the mean from each observation : $x_i - \mu$
#### 3- Square each of the resulting observations : ($x_i - \mu)^2$
#### 4- Add these squared results together.
#### 5- Divide this total by the number of observations n (in case of population).

```python
# Calculate Variance using Traditional way

data = np.array([ 4,  2,  5, 10,  3, 11, 20, 15])

# 1- Calculate data mean
x_mean = data.mean()
print('the mean of data is :', x_mean) # 8.75
# 2- Calculate data length
n = len(data) 

# 3- Calculating square diff
data_diff = [((x - x_mean ) ** 2) for x in data]

# Calculating summ

data_sum = sum(data_diff)

# Dividing by number of points (n)

var = data_sum / n
print(var)

# Calculate Variance using numpy in one step
print(np.var(data))
```

* Note : Variance unit is the square of the data unit, for ex if data is about temperature degrees the variance unit would be $C^2$ instead of C

## **4- Standard Deviation**
#### Positive square root of variance that also measures how far a group of data points are far from the mean but has the same unit of data.
# $\sigma = \sqrt{\frac{\sum_{i=1}^n (x_i - \mu)^2}{n}}$

```python
# Calculate the standard deviation from the variance

print((var) ** 0.5)

# Or using numpy
print(data.std())
```

## **Correlation**
#### A measure that descripes to what extent two variables are linearly related.

image.png

* A correlation of –1 indicates a perfect negative correlation, meaning that as one variable goes up, the other goes down.
* A correlation of +1 indicates a perfect positive correlation , meaning that as one variable goes up, the other goes up together.
* A correlation of zero indicates that there is no relationship between the variables.

image.png

```python
# Calculate Pearson Coefficient for Correlation

arr_1 = np.arange(1, 11)
arr_2 = np.array([12, 15, 20, 30, 47, 50, 88, 90, 95, 100])

corr_two_arr = np.corrcoef(arr_1, arr_2)
print(corr_two_arr)                                            # Result shows a strong positive correlation between the two arrays
```

image.png

## Normal Distribution

### Formula
# $f(x) = \frac{1}{\sigma\sqrt{2\pi}} e^{-\frac{1}{2}(\frac{x-\mu}{2\sigma})}$

### Characteristics

image.png

### Empirical Rule
#### The empirical rule states that in a normal distribution, virtually all observed data will fall within three standard deviations of the mean. Under this rule, 68% of the data will fall within one standard deviation, 95% within two standard deviations, and 99.7% within three standard deviations from the mean

image.png

## Standard Normal Distribution

### Formula
# $z_{score} = \frac{x - \mu}{\sigma}$

### Characteristics

image.png

image.png

### Why use Z-Score ?
#### One of the benefits of Z-Score is that it can be used to compare different normally distributed datasets

### Example

image.png

```python
# First Exam
first_grade = 80
first_mean = 70
first_std = 15

# Second Exam
second_grade = 75
second_mean = 64
second_std = 12

# Calculate Z-Score
z_score_first = (first_grade - first_mean) / first_std
z_score_second = (second_grade - second_mean) / second_std

print(f"The student's Z-Scores in both exams respectively : {round(z_score_first, 2)}, {round(z_score_second, 2)}")
```

#### Conclusion
* The Student performed better in second exam relative to his colleagues
