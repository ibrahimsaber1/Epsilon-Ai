### Applting Functions on Dataframes

```python
df = pd.DataFrame({'col1':[1,5,3,2,4],'col2':[444,222,666,444,555],'col3':['abc','xyz','ijklmn','xyz', 'defgh']})
df
```

```python
# Create Radius column from (col1 + 20)

df['radius'] = df['col1'] + 20
df
```

```python
# Create Circle Area column

def circ_area(x):

    return 3.14 * (x ** 2)

df['circ_area'] = df['radius'].apply(circ_area)
```

```python
# Create multiple columns from function

def circ_area_perimeter(x):

    return pd.Series([3.14 * (x ** 2), 2 * 3.14 * x])
```

```python
df[['new_rea', 'new_permieter']] = df['radius'].apply(circ_area_perimeter)
```

```python
df
```

```python
# Create Circle Area column (Using lambda)

df['circ_area2'] = df['radius'].apply(lambda x : 3.14 * (x ** 2))
```

```python
df
```

```python
# Apply built in functions with apply()

df['col3'].apply(len)
```

```python
df
```

### Merge

```python
left = pd.DataFrame({'key': ['K0', 'K1', 'K2', 'K3'],
                     'A': ['A0', 'A1', 'A2', 'A3'],
                     'B': ['B0', 'B1', 'B2', 'B3']})
   
right = pd.DataFrame({'key': ['K0', 'K1', 'K2', 'K3', 'k4'],
                          'C': ['C0', 'C1', 'C2', 'C3', 'C4'],
                          'D': ['D0', 'D1', 'D2', 'D3', 'D4']})
```

```python
left
```

```python
right
```

```python
pd.merge(left, right, on= 'key', how= 'outer')
```
