## Rename Columns in a suitable form

```python
df.columns = df.columns.str.lower()
df.columns
```

```python
df.columns = df.columns.str.strip().str.replace(' ', '_')
df.columns
```

```python
df
```

```python
int('1990')
```

```python
if 'I' in '(I) (2020)':
    print('There is a bracket')
```

```python
def clean_year(x):

    if 'I' in x:
        return int(x.split()[1].replace('(', '').replace(')', ''))
        
    else:
        return int(x.replace('-', ''))

df['year'] = df['year'].apply(clean_year)
df['year']
```

```python
df['year'].unique()
```

```python
df
```

```python
df['runtime'] = df['runtime'].apply(lambda x : int(x.replace(' min', '')))
df['runtime']
```
