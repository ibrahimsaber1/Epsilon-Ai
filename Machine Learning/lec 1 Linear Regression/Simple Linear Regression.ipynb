```python
import numpy as np
import pandas as pd
import plotly.express as px
```

```python
df = pd.read_csv('Salary_dataset.csv', index_col= 0)
df.head()
```

```python
x_bar = df.YearsExperience.mean()
y_bar = df.Salary.mean()

theta_1 = sum( (df.YearsExperience - x_bar) * (df.Salary - y_bar) ) / sum((df.YearsExperience - x_bar) ** 2)
theta_0 = y_bar - (theta_1 * x_bar)

print(theta_0, theta_1)
```

```python
from sklearn.linear_model import LinearRegression

le = LinearRegression()

le.fit(df[['YearsExperience']], df['Salary'])
```

```python
le.intercept_
```

```python
le.coef_
```

```python
df[['YearsExperience']].ndim
```

```python
np.array([[3.4]]).ndim
```

```python
le.predict(np.array([[3.4]]))
```

```python
salary_predicted_ols = le.predict(df[['YearsExperience']])
```

```python
df['salary_predicted_ols'] = salary_predicted_ols
```
