```python
df.drop(['Ad Topic Line', 'City','Timestamp'], axis=1 , inplace=True)
```

```python
df.info()
```

```python
# data preprocising :-
x = df.drop('Clicked on Ad' ,axis=1)
y = df['Clicked on Ad']
```

```python
y.value_counts()
```

```python
from sklearn.model_selection import train_test_split

x_train , x_test , y_train , y_test = train_test_split(x,y, test_size=0.2,
                                                        random_state=42,
                                                        stratify=y
                                                        )
```

```python
rc_cols = x_train.columns.drop(['Country', 'Male'])
rc_cols
```

```python
from sklearn.preprocessing import RobustScaler

rc = RobustScaler()
x_train[rc_cols] = rc.fit_transform(x_train[rc_cols])
x_test[rc_cols] = rc.transform(x_test[rc_cols])
```

```python
from category_encoders import BinaryEncoder

bi = BinaryEncoder()

bi_train = bi.fit_transform(x_train[['Country']],)
bi_test = bi.transform(x_test[['Country']])
```

```python
type(bi_train)
```

```python
x_train= pd.concat([x_train, bi_train] , axis=1).drop('Country', axis=1)
x_test = pd.concat([x_test, bi_test] , axis=1).drop('Country', axis=1)
```

```python
x_test
```

```python
# ML 

from sklearn.linear_model import LogisticRegression

lr = LogisticRegression(penalty=None)

lr.fit(x_train, y_train)

test_score = lr.score(x_test , y_test)
train_score = lr.score(x_train , y_train)

print("test_score ===>", train_score * 100)
print("train_score ===>", test_score * 100)
```

```python
# ML 

from sklearn.linear_model import LogisticRegression

lr = LogisticRegression(penalty='l2',
                        C=1)

lr.fit(x_train, y_train)

test_score = lr.score(x_test , y_test)
train_score = lr.score(x_train , y_train)

print("test_score ===>", train_score * 100)
print("train_score ===>", test_score * 100)
```

```python
# ML 

from sklearn.linear_model import LogisticRegression

lr = LogisticRegression(penalty='l2',
                        C=0.5)

lr.fit(x_train, y_train)

test_score = lr.score(x_test , y_test)
train_score = lr.score(x_train , y_train)

print("test_score ===>", train_score * 100)
print("train_score ===>", test_score * 100)
```
