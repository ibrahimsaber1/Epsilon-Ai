# Linear Regression

* The goal of a regression model is to build a mathematical equation that defines y as a function of the x variables.
* Next, this equation can be used to predict the outcome (y) based on new values of the predictor variables (x).
* Regression is a set of machine learning methods that allow us to predict a continuous outcome variable (y) based on the value of one or multiple predictor variables (x).
* Regression follows under the supervised machine learning type.

## 1) Simple Linear Regression
##### Is used to estimate the relationship between two quantitative variables.
##### Based on the idea that the relationship between two variables can be explained by the following formula (straight line equation):
# $y = \theta_{0} + \theta_{1}x$
* y : Output Variable
* x : Input Variable
* $\theta_{0}$ : Intercept
* $\theta_{1}$ : coefficient of x

### Finding Best Fit Line
* Why a line : To predict future values using line equation for ex : Price = 1000 Area + 20
* How to find best fit line ? : By finding the line with least square error (leasts sum of square distances from actual data points)
# $\text{RSS} = \sum_{i=1}^{n} (\hat{y}_i - y_i)^2 = \sum_{i=1}^{n} ((\theta_{0} + \theta_{1}x_i) - y_i)^2$

#### 2) **Optimization Solution Gradient Descent**

Gradient Descent is an optimization algorithm used to minimize a cost function by iteratively updating parameters in the direction of the steepest descent.

---

## **1. Initialize Parameters**

Start with initial values (often random or zero) for the model parameters, e.g.:

- For Simple linear regression: $( \theta_0, \theta_1)$

---

## **2. Compute the Cost Function**

Use a cost function to evaluate how well the model fits the data.  
For linear regression, the Mean Squared Error (MSE) cost function is:

# $J(\theta) = \frac{1}{2n} \sum_{i=1}^{n} \left(\hat{y} - y_i \right)^2$

where:

- (n) : Averages the total squared error over all training samples, Without this the cost would scale with the size of the dataset.  
- ($\frac{1}{2}$) : Cancels the 2 in the derivative, making gradient expressions neat.

---

## **3. Repeat Until Convergence**

### **a. Compute the Gradient**

Calculate the partial derivatives of the cost function with respect to each parameter:

# $\frac{\partial J}{\partial \theta_1} = \frac{1}{n} \sum_{i=1}^{n} \left[(\theta_{0} + \theta_{1}x_i) - y_i \right] x_i$

# $\frac{\partial J}{\partial \theta_0} = \frac{1}{n} \sum_{i=1}^{n} \left[(\theta_{0} + \theta_{1}x_i) - y_i \right]$

### **b. Update the Parameters**

Update each parameter using the learning rate ( $\alpha$ ):

# $\theta_1new = \theta_1old - \alpha \cdot \frac{\partial J}{\partial \theta_1}$

# $\theta_0new = \theta_0old - \alpha \cdot \frac{\partial J}{\partial \theta_0}$

---

## **4. Check for Convergence**

Stop the iterations when:

- The cost function $(J(\theta_0, \theta_1))$ stops decreasing significantly, or
- A maximum number of iterations is reached

---

## Notes on Gradient Descent Variants

- **Batch Gradient Descent**: Uses all training data in each iteration.
- **Stochastic Gradient Descent (SGD)**: Updates parameters using one data point at a time.
- **Mini-batch Gradient Descent**: Uses small subsets (batches) of data for each update.
