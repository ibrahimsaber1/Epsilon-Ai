```python
import pandas as pd
import numpy as np
import plotly.express as plt
```

```python
df = pd.read_csv('./Salary_dataset.csv', index_col=0)
df.head()
# df.shape
```

```python
x_bar = df.YearsExperience.mean()
y_bar = df.Salary.mean()
print('x_bar ==>', x_bar)
print('y_bar ==>', y_bar)
```

```python
theta_1 = sum((df.YearsExperience - x_bar) * (df.Salary - y_bar)) /sum((df.YearsExperience - x_bar)**2)
theta_1
```

```python
theta_0 = y_bar - (theta_1 * x_bar)
theta_0
```

```python
from sklearn.linear_model import LinearRegression

lr = LinearRegression()

lr.fit(df[['YearsExperience']], df[['Salary']])
```
